```python
r = cq.Workplane("front").circle(2.0)  # make base
r = r.pushPoints(
    [(1.5, 0), (0, 1.5), (-1.5, 0), (0, -1.5)]
)  # now four points are on the stack
r = r.circle(0.25)  # circle will operate on all four points
result = r.extrude(0.125)  # make prism
show(result)
```

```python
################ The Valve system ################


import cadquery as cq

# Outer radius and wall thickness
outer_r = 1
thickness = 0.2
inner_r = outer_r - thickness

# === Build outer solid ===
result = (
    cq.Workplane("front")
    .circle(outer_r)
    .extrude(10)  # Main cylinder
    .copyWorkplane(
        cq.Workplane("right", origin=(-5, 0, 0))
    )
    .circle(outer_r)
    .extrude(10)  # Perpendicular cylinder
)

# === Hollow main cylinder ===
result = result.cut(
    cq.Workplane("front")
    .circle(inner_r)
    .extrude(10)
)

# === Hollow perpendicular cylinder ===
result = result.cut(
    cq.Workplane("right", origin=(-5, 0, 0))
    .circle(inner_r)
    .extrude(10)
)

# Show it
show(result)
```

```python
import cadquery as cq

# === PARAMETERS ===
outer_radius1 = 25
outer_radius2 = 15
height1 = 4
height2 = 70
hole_radius = 10  # radius of the center hole
hole_height = height1 + height2  # make sure hole goes through both cylinders

# === CYLINDER 1 (Vertical, Larger) ===
cyl1 = cq.Workplane("XY").circle(outer_radius1).extrude(height1)

# === CYLINDER 2 (Vertical, Smaller, Concentric) ===
cyl2 = cq.Workplane("XY").circle(outer_radius2).extrude(height2).translate((0, 0, height1))

# === COMBINE BOTH ===
assembly = cyl1.union(cyl2)

# === SUBTRACT CENTER HOLE ===
hole = cq.Workplane("XY").circle(hole_radius).extrude(hole_height)
assembly = assembly.cut(hole)

# === SHOW RESULT ===
show(assembly)
```
